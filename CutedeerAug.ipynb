```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_datagen = ImageDataGenerator(
rotation_range=40,
width_shift_range=0.2,
height_shift_range=0.2,
rescale=1./255,
shear_range=0.2,
zoom_range=0.2,
horizontal_flip=True)
```

```python
from keras.preprocessing.image import load_img, img_to_array
# Load the image
img = load_img("/content/sample_data/Cute_Deer.jpeg")
img
```

```python
# Convert the image to a NumPy array
x = img_to_array(img)
x
```

```python
# Reshape the array
x = x.reshape((1,) + x.shape)
```

```python
i = 0
for x in train_datagen.flow(x, batch_size=1, save_to_dir='/content/sample_data/Cute_Deer_Aug', save_format='jpeg'):
      i += 1
      if i > 10:
         break
```
